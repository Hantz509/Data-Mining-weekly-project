## Dataset 1: Retail Market Basket Analysis (Discrete Quantities)

**Scenario:** You are analyzing transaction records from a local bakery and coffee shop. Unlike standard Apriori, you have the exact quantity of each item purchased by the customers. Your goal is to find significant associations in purchasing behavior based on *quantities*, bypassing the vector normalization step since these are discrete integer transactions.

**Suggested Parameters:**
* **Minimum Support (`min_sup`):** `3`
* **Target:** Identify all frequent 2-itemsets and extract at least two association rules with their Confidences.

```python
import pandas as pd

# Dataset 1: Transactions with exact item quantities
data_transactions = {
    'Espresso': [2, 0, 1, 3, 2],
    'Croissant': [2, 1, 2, 0, 3],
    'Muffin': [0, 3, 0, 2, 1],
    'Latte': [0, 2, 1, 0, 1],
    'Tea': [1, 0, 2, 1, 0]
}
transactions = ['T1', 'T2', 'T3', 'T4', 'T5']
df_retail = pd.DataFrame(data_transactions, index=transactions)

print("Dataset 1: Retail Transactions")
display(df_retail)
```
