### 📝 Convention Used

We follow the **standard GSP definition** (Srikant & Agrawal, 1996; Han & Kamber, *Data Mining: Concepts and Techniques*):

> A sequence $S = \langle s_1, s_2, \ldots, s_n \rangle$ **contains** subsequence $T = \langle t_1, t_2, \ldots, t_m \rangle$ if there exist indices $i_1 < i_2 < \cdots < i_m$ such that $t_k = s_{i_k}$ for all $k$.

Elements must appear **in order**, but **not necessarily contiguously**. For example, $\langle A, D \rangle$ is contained in $\langle A, B, D, E \rangle$ even though $A$ and $D$ are not adjacent.

**Support count** of a pattern = number of input sequences that contain it as a subsequence.

### ✅ Task 1 Answers (min_support = 2)

#### Step 1 — Frequent 1-sequences ($L_1$)

Scan each event across the 4 sequences:

| 1-sequence | S1 ⟨A,B,D,E⟩ | S2 ⟨B,C⟩ | S3 ⟨A,B,C,D⟩ | S4 ⟨A,C⟩ | Support | Frequent? |
|------------|:---:|:---:|:---:|:---:|:---:|:---:|

| ⟨B⟩ | ✓ | ✓ | ✓ | – | **3** | ✅ |

| ⟨E⟩ | ✓ | – | – | – | **1** | ❌ (pruned) |

$$L_1 = \{\langle A \rangle, \langle B \rangle, \langle C \rangle, \langle D \rangle\} \quad \text{(4 patterns)}$$

#### Step 2 — Candidate generation $C_2$ and support counting

From $L_1 = \{A, B, C, D\}$, generate all $4 \times 4 = 16$ ordered pairs as candidates. Then count how many sequences contain each pair **in the given order** (not necessarily adjacent):

| 2-sequence | S1 | S2 | S3 | S4 | Support | Frequent? |
|------------|:---:|:---:|:---:|:---:|:---:|:---:|
| ⟨A,B⟩ | ✓ (A→B) | – | ✓ (A→B) | – | **2** | ✅ |
| ⟨A,C⟩ | – | – | ✓ (A→C) | ✓ (A→C) | **2** | ✅ |
| ⟨A,D⟩ | ✓ (A→D) | – | ✓ (A→D) | – | **2** | ✅ |
| ⟨B,C⟩ | – | ✓ (B→C) | ✓ (B→C) | – | **2** | ✅ |
| ⟨B,D⟩ | ✓ (B→D) | – | ✓ (B→D) | – | **2** | ✅ |
| ⟨C,D⟩ | – | – | ✓ (C→D) | – | **1** | ❌ |
| ⟨B,A⟩, ⟨C,A⟩, ⟨C,B⟩, ⟨D,A⟩, ⟨D,B⟩, ⟨D,C⟩ | all 0 | | | | **0** | ❌ |
| ⟨A,A⟩, ⟨B,B⟩, ⟨C,C⟩, ⟨D,D⟩ (no repeats in dataset) | | | | | **0** | ❌ |

$$L_2 = \{\langle A,B \rangle, \langle A,C \rangle, \langle A,D \rangle, \langle B,C \rangle, \langle B,D \rangle\} \quad \text{(5 patterns)}$$

#### Step 3 — Candidate generation $C_3$, prune, and count

**GSP join rule:** $\langle s_1, s_2 \rangle \bowtie \langle s_2, s_3 \rangle \rightarrow \langle s_1, s_2, s_3 \rangle$ (last item of first = first item of second).

Joining $L_2$:
- $\langle A,B \rangle \bowtie \langle B,C \rangle \rightarrow \langle A,B,C \rangle$
- $\langle A,B \rangle \bowtie \langle B,D \rangle \rightarrow \langle A,B,D \rangle$
- $\langle A,C \rangle$, $\langle A,D \rangle$, $\langle B,C \rangle$, $\langle B,D \rangle$: no frequent pair starts with $C$ or $D$ → no joins.

**Prune step** (all length-2 subsequences must be in $L_2$):
- $\langle A,B,C \rangle$: subseqs ⟨A,B⟩ ✓, ⟨A,C⟩ ✓, ⟨B,C⟩ ✓ → **keep**
- $\langle A,B,D \rangle$: subseqs ⟨A,B⟩ ✓, ⟨A,D⟩ ✓, ⟨B,D⟩ ✓ → **keep**

**Support counting:**

| 3-sequence | S1 ⟨A,B,D,E⟩ | S2 ⟨B,C⟩ | S3 ⟨A,B,C,D⟩ | S4 ⟨A,C⟩ | Support | Frequent? |
|------------|:---:|:---:|:---:|:---:|:---:|:---:|
| ⟨A,B,C⟩ | A<B but no C | – | ✓ A→B→C | – | **1** | ❌ |
| ⟨A,B,D⟩ | ✓ A→B→D | – | ✓ A→B→D | – | **2** | ✅ |

$$L_3 = \{\langle A,B,D \rangle\} \quad \text{(1 pattern)}$$

#### Step 4 — $C_4$

Only one 3-sequence is frequent, so no joins are possible.
$$L_4 = \emptyset$$

### 📋 Task 1 — Final Output

| Length | Frequent Sequential Patterns (min_sup = 2) | Count |
|:------:|--------------------------------------------|:-----:|
| 1 | ⟨A⟩:3, ⟨B⟩:3, ⟨C⟩:3, ⟨D⟩:2 | **4** |
| 2 | ⟨A,B⟩:2, ⟨A,C⟩:2, ⟨A,D⟩:2, ⟨B,C⟩:2, ⟨B,D⟩:2 | **5** |
| 3 | ⟨A,B,D⟩:2 | **1** |
| **Total** | | **10** |

````python
# Optional: verification with Python
from itertools import product

sequences = {
    'S1': ['A','B','D','E'],
    'S2': ['B','C'],
    'S3': ['A','B','C','D'],
    'S4': ['A','C']
}

def contains_subseq(seq, pat):
    """Does seq contain pat as an ordered subsequence?"""
    i = 0
    for x in seq:
        if i < len(pat) and x == pat[i]:
            i += 1
    return i == len(pat)

def support(pat):
    return sum(contains_subseq(s, pat) for s in sequences.values())

def gsp(min_sup, max_len=4):
    items = sorted({x for s in sequences.values() for x in s})
    # L1
    L = [[(it,) for it in items if support((it,)) >= min_sup]]
    print(f'L1: {L[0]}  counts: {[support(p) for p in L[0]]}')
    k = 2
    while L[-1] and k <= max_len:
        # GSP join on size-k
        prev = L[-1]
        candidates = set()
        for a, b in product(prev, repeat=2):
            if a[1:] == b[:-1]:               # join rule
                cand = a + (b[-1],)
                # prune: all (k-1)-subsequences must be in prev
                if all(tuple(cand[:i]+cand[i+1:]) in set(prev) for i in range(len(cand))):
                    candidates.add(cand)
        Lk = [c for c in sorted(candidates) if support(c) >= min_sup]
        print(f'L{k}: {Lk}  counts: {[support(p) for p in Lk]}')
        L.append(Lk)
        k += 1
    return L

print('=== min_support = 2 ===')
_ = gsp(2)
````

---

**1. Why do fewer patterns pass at higher support thresholds?**

Raising `min_support` raises the bar a pattern must clear. With only 4 input sequences, `min_sup = 3` means a pattern must occur in **75% of customers** to qualify — that is a strong consensus requirement. The **Apriori (anti-monotone) property** also amplifies this effect: when an item like ⟨D⟩ falls below the threshold, *every* longer pattern containing D is automatically pruned, regardless of how often it would have appeared. Higher thresholds therefore prune both directly (items themselves fail) and indirectly (their supersets get eliminated cascade-style).

**2. Which sequences disappear when you raise min_support from 2 to 3?**

Going from min_sup=2 → 3:
- $L_1$ loses ⟨D⟩  (support 2 → infrequent)
- $L_2$ loses **all five** patterns: ⟨A,B⟩, ⟨A,C⟩, ⟨A,D⟩, ⟨B,C⟩, ⟨B,D⟩ (each had support 2)
- $L_3$ loses ⟨A,B,D⟩

Total patterns drops from **10 → 3**, a 70% reduction.

**3. What does this tell you about customer behavior?**

- The base events A, B, C are **universally popular** (≥75% of customers touch them) — these are the "core touchpoints" of the site/app.
- Event D is a **secondary touchpoint** — present for half of customers, gone at the strict threshold.
- Event E is a **fringe event** — only 1 customer ever reaches it (perhaps a deep funnel step like checkout success).
- No multi-step **journey** is shared by ≥75% of customers, which means individual paths through the site are **diverse**. Marketing/UX insight: there is no single dominant funnel here; customers take many different routes. At a relaxed threshold (50%) several recurring 2-step paths *do* emerge (A→B, A→C, A→D, B→C, B→D), suggesting the dataset has a soft "prefix" pattern where A and B tend to be entry events.
- The 3-sequence ⟨A,B,D⟩ at min_sup=2 hints at a possible mini-funnel A → B → D that's worth examining with a larger sample.

**Threshold-setting takeaway:** With only 4 sequences, min_sup=3 is *over-restrictive*; min_sup=2 (50%) gives a richer, more usable pattern set. In real datasets with thousands of sequences, absolute support is usually replaced by **relative support** (e.g., 1% or 5%) for the same reason.

### 📊 Comparison: GSP vs. Apriori on This Dataset

| Level | GSP (min_sup=2) | Apriori (min_sup=2) | Same count? |
|:-----:|-----------------|---------------------|:-----------:|
| 1 | ⟨A⟩, ⟨B⟩, ⟨C⟩, ⟨D⟩ | {A}, {B}, {C}, {D} | ✅ 4 = 4 |
| 2 | ⟨A,B⟩, ⟨A,C⟩, ⟨A,D⟩, ⟨B,C⟩, ⟨B,D⟩ | {A,B}, {A,C}, {A,D}, {B,C}, {B,D} | ✅ 5 = 5 |
| 3 | ⟨A,B,D⟩ | {A,B,D} | ✅ 1 = 1 |

#### Why the counts match here (a special case, not a general rule)

On this particular dataset the totals coincide because the events in every sequence happen to appear in roughly the same alphabetical order (A → B → C → D). Therefore, whenever two items co-occur in a transaction, they co-occur in **exactly one direction** in the original sequence. There is no "backward" version like ⟨B, A⟩ or ⟨D, C⟩ to split the support across two GSP patterns.

#### Why Apriori ≥ GSP in general

Conceptually, **flattening always equals or increases pattern counts**:
- Itemset $\{X, Y\}$ is frequent in Apriori iff $\text{sup}(\langle X,Y \rangle) + \text{sup}(\langle Y,X \rangle) \ge$ min_sup (treating co-occurrences either way).
- GSP must clear the bar with each direction *separately*.

**Constructed counter-example.** Add a fifth sequence $S_5 = \langle D, B, A \rangle$:
- Apriori: $\{A, B\}$ now has support **3** — still frequent.
- GSP: $\langle A, B \rangle$ stays at **2**, and a new $\langle B, A \rangle$ appears with support **1** — both fall under the joint threshold the moment we set min_sup = 3.

So Apriori would still report the pattern, while GSP would lose it. This is the essential informational trade-off: **GSP preserves order at the cost of fragmenting support**.

#### When to use which

| Use **GSP / sequential mining** when… | Use **Apriori / itemset mining** when… |
|---------------------------------------|----------------------------------------|
| The order of events carries meaning (clickstreams, customer journeys, event logs, DNA/protein motifs) | Order is irrelevant or unknown (market-basket co-occurrence) |
| You want to discover funnels and temporal flows | You want to discover bundles / cross-sell pairs |
| Timestamps or session order are reliable | Data already comes as unordered sets |

**Bottom line:** GSP carries **more information** (order), Apriori is **more permissive** (any co-occurrence). On a perfectly time-aligned dataset like ours they may agree numerically, but in any realistic clickstream they will diverge.

````python
# Optional: verification of Apriori
from itertools import combinations

transactions = {
    'T1': {'A','B','D','E'},
    'T2': {'B','C'},
    'T3': {'A','B','C','D'},
    'T4': {'A','C'},
}

def sup_set(itemset):
    return sum(itemset.issubset(t) for t in transactions.values())

def apriori(min_sup):
    items = sorted({x for t in transactions.values() for x in t})
    L = [[frozenset([it]) for it in items if sup_set({it}) >= min_sup]]
    print(f'L1: {[set(s) for s in L[0]]}  counts: {[sup_set(s) for s in L[0]]}')
    k = 2
    while L[-1]:
        seen = set()
        cands = set()
        for a, b in combinations(L[-1], 2):
            u = a | b
            if len(u) == k:
                # prune
                if all(frozenset(c) in set(L[-1]) for c in combinations(u, k-1)):
                    cands.add(u)
        Lk = [c for c in sorted(cands, key=lambda s: sorted(s)) if sup_set(c) >= min_sup]
        if not Lk: break
        print(f'L{k}: {[set(s) for s in Lk]}  counts: {[sup_set(s) for s in Lk]}')
        L.append(Lk); k += 1
    return L

print('=== Apriori, min_support = 2 ===')
_ = apriori(2)
````
